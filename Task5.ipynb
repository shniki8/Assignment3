```python
#COPIED:
import numpy as np
from scipy.stats import norm

def analyse_ab_test(label, purchases_a, visitors_a, purchases_b, visitors_b, alpha=0.05):
    rate_a = purchases_a / visitors_a
    rate_b = purchases_b / visitors_b
    difference = rate_b - rate_a
    relative_increase = difference / rate_a

    # Pooled standard error for testing H0: p_A = p_B
    pooled_rate = (purchases_a + purchases_b) / (visitors_a + visitors_b)
    se_test = np.sqrt(
        pooled_rate * (1 - pooled_rate) * (1 / visitors_a + 1 / visitors_b)
    )
    z = difference / se_test
    p_value = 2 * norm.sf(abs(z))

    # Unpooled standard error for the confidence interval
    se_interval = np.sqrt(
        (rate_a * (1 - rate_a) / visitors_a) +
        (rate_b * (1 - rate_b) / visitors_b)
    )
    critical_value = norm.ppf(1 - alpha / 2)
    lower = difference - critical_value * se_interval
    upper = difference + critical_value * se_interval

    print(f"\n{label}")
    print("-" * len(label))
    print(f"Observed conversion rate A: {rate_a:.2%}")
    print(f"Observed conversion rate B: {rate_b:.2%}")
    print(f"Estimated difference (B-A): {100 * difference:.2f} percentage points")
    print(f"Relative increase over A: {relative_increase:.2%}")
    print(f"Test statistic z: {z:.3f}")
    print(f"Two-sided p-value: {p_value:.4g}")
    print(f"Reject H0 at alpha {alpha}: {p_value < alpha}")
    print(f"95% confidence interval for p_B-p_A: [{100 * lower:.3f}, {100 * upper:.3f}] percentage points")
    print(f"Confidence interval width: {100 * (upper - lower):.3f} percentage points")

# Original experiment
analyse_ab_test(
    "Original experiment",
    purchases_a=500, visitors_a=5000,
    purchases_b=575, visitors_b=5000
)

# Larger experiment
analyse_ab_test(
    "Larger experiment, same observed conversion rates",
    purchases_a=5000, visitors_a=50000,
    purchases_b=5750, visitors_b=50000
)
```

P-value is far smaller than 0.05 which means we have to reject H0 at the chosen significance level, which is also shown by the fact that the confidence interval doesn't contain 0, which is our H0 for p_A - p_B since they are the same according to H0.
