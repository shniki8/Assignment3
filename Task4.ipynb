```python
from random import sample

import numpy as np
from matplotlib.pyplot import ylabel, xlabel, title

rng = np.random.default_rng(0)
population = rng.exponential(scale=10, size=100_000) # minutes
```

```python
import seaborn as sns
ax = sns.histplot(data=population, bins=50)
ax.set(ylabel="Amount of trips", xlabel="Trip duration", title = "Distribution of trip durations")
print("Mean: ",population.mean())
print("Standard Deviation: ",population.std())
print("Median: ",population[50000])
```

```python
means_dict = {}
for n in [5, 30, 200]:
    samples = rng.choice(population, size=(5000, n))
    sample_means = samples.mean(axis=1)
    means_dict[f"n = {n}"] = sample_means
ax = sns.histplot(data=means_dict, bins=50, palette=["#4c72b0", "#dd8452", "#c44e52"])
ax.set(ylabel="Amount", xlabel="Sample Mean", title="Sampling distribution of the mean")
```

Distribution centers around real mean as showcased by the central limit theorem.
