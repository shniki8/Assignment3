```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

flights = pd.read_csv("data/flights.csv")
flights = flights.groupby(["day", "year", "month"]).agg(count=("id", "count")).reset_index()
plt.figure(figsize=(15, 5))
flights["date"] = pd.to_datetime(flights[["year", "month", "day"]])
ax = sns.lineplot(data=flights, x="date", y="count", color="pink")
ax.set(ylabel="Number of flights", xlabel="Date", title="Flights per day")
```


Could not plot number of passangers/year since the dataset only contains data for 2013 and no ticketing information but lineplot would've been the best to see the trend. As an alternative I plotted the flights per day which shows the weekly cycles with a dip each

```python
tips = pd.read_csv("data/tip.csv")
print(tips["day"].unique())
tips = tips.groupby("day").aggregate(average=("tip","mean"))
ax = sns.barplot(tips, x="day", y="average", color="pink", order=["Thur", "Fri", "Sat", "Sun"])
ax.set(title="Mean tips amount per day of the week", ylabel="Mean amount", xlabel="Day")
```

I used a bar plot since we compare numeric quantities to each other.

```python
titanic = pd.read_csv("data/Titanic-Dataset.csv")
titanic = titanic["Age"]
ax = sns.histplot(titanic, color="crimson", binwidth=1)
ax.set(ylabel="Number of passengers", xlabel="Age", title="Age distribution of titanic passengers")
titanic.describe()
```

I used a histogram since we want to visualize a distribution.
50% of passengers were between 20 and 38.

```python
penguins = pd.read_csv("data/penguins.csv")
ax = sns.scatterplot(penguins, x="flipper_length_mm", y="body_mass_g")
ax.set(title="Flipper length in relation to body mass", ylabel="Body mass", xlabel="Flipper length")
```

I chose a scatter plot since it is the ideal tool to show the relation between the two numerical features.
We can see that a higher the body mass relates to a higher flipper length (which is logical).
